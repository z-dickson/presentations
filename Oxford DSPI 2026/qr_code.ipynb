```python
import qrcode

# URL you want the QR code to point to
url = "https://z-dickson.github.io/assets/dickson_energy_prices.pdf"   # change if needed

qr = qrcode.QRCode(
    version=None,  # let library pick the right size
    error_correction=qrcode.constants.ERROR_CORRECT_M,
    box_size=10,   # size of each "square"
    border=4,      # border width (in boxes)
)

qr.add_data(url)
qr.make(fit=True)

img = qr.make_image(fill_color="black", back_color="white")

# Save a high-resolution PNG for slides
img.save("qr_website.png", dpi=(300, 300))
```
